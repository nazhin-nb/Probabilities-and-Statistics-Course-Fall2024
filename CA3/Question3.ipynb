```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom, norm
```

part1:

```python
n = 270
p = 0.3
X = np.arange(n+1)
binomial_pmf = binom.pmf(X, n, p)
plt.bar(X, binomial_pmf)
```

part2:

```python
mean = n * p
std = np.sqrt(n * p * (1 - p))
X_normal = (X - mean) / std
normal_pdf = norm.pdf(X_normal, 0, 1)
plt.bar(X_normal, binomial_pmf)
plt.plot(X_normal, normal_pdf)
```

part3:

```python
plt.bar(X_normal, binomial_pmf)
plt.plot(X_normal, normal_pdf)
```

part5:

از رابطه ریمان گفته شده استفاده کردیم تا رابطه ها به صورت زیر تصحیح شوند

```python
a = 270 * 0.3 - 4 * std
b = 270 * 0.3 + 4 * std
N = 1000
x = np.linspace(a, b, N)
f = norm.pdf(x, mean, std)
riemann_sum =  (b-a) / N * np.sum(f)
scale = 1 / riemann_sum
plt.bar(X, binomial_pmf)
plt.plot(x, f * scale, color='red')
print(scale, std)
```

دو نمودار نرمال و دوجمله ای بر هم منطبق شده اند زیرا از روابط ضریب اسکیل استفاده کردیم
