```r
exponential_distribution <- function(n, lambda)
{
    input <- rexp(n, rate = lambda)
    return (input)
}
```

```r
Time <- function(exp_distribution, arrival,maxtime)
{
  result <- exp_distribution [exp_distribution > arrival & exp_distribution < maxtime]
  entering <- (result - arrival)
  return(entering)
}
```

```r
simulate <- function(time)
{
  lambda <- 1 / 15
  n <- 1000*100
  exp_distribution_ans <- exponential_distribution(n,lambda)
  time <- Time(exp_distribution_ans-12, 12, 8*60)
  simulation_probability <- mean(time <= 15)
  return(simulation_probability)
}

theorical <- function(wait_time)
{
  lambda <- 1/15
  theorical_answer <- 1-exp(-lambda * (wait_time - 12))
  return(theorical_answer)
}
simulate(15)
theorical(15)
```
