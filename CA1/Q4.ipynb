part1:

```r
set.seed(530)
n <- 10**6
x <- runif(n,0,1)
x
```

part2:

```r
y <- -2 * log(x)
y
```

part3:

```r
hist(y, breaks = 25, probability = TRUE, col = "red",xlab = "Y",main = "Simulated and Theoretical Comprasion:")
curve(0.5 * exp(-x/2), col="black",lwd = 1, add = TRUE)
```

part4:

```r
set.seed(530)
n <- 10**6
U1 <- runif(n,0,1)
U1
U2 <- runif(n,0,1)
U2
```

part5:

```r
#ba estefade az rahnamayi sosal taghir moteghayer dadim
R <- sqrt(-2*log(U1))
theta <- 2 * pi * U2
Z1 <- R * cos(theta)
Z1
Z2 <- R * sin(theta)
Z2
```

part6:

```r
par(mfrow = c(1, 2))
hist(Z1, probability = TRUE, col ="red", main = "Z1 Distribution",breaks = 20, xlab = "Z1", xlim = c(-5, 5),ylim = c(0,0.5))
curve(dnorm(x), col = "blue", lwd = 2, add = TRUE)
hist(Z2, probability = TRUE, col = "yellow", main = "Z2 Distribution",breaks=20, xlab = "Z2", xlim = c(-5, 5), ylim = c(0,0.5))
curve(dnorm(x), col = "black", lwd = 2, add = TRUE)
```
