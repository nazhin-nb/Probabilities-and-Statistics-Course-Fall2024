## Libraries

```python
import re
import string
import csv
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix
```

## Part A

```python
EMAILS_FILE_ADDRESS = 'emails.csv'
global EMAILS_FILE_ADDRESS

def cleanText(text):
    cleanedText = re.sub(r'\d+', '', text[0]).lower()
    cleanedText = cleanedText.translate(str.maketrans('', '', string.punctuation))
    output = {"text": cleanedText, "spam": text[1]}
    return output

def preprocessEmails():
    cleanedEmails = []
    csvfile = open(EMAILS_FILE_ADDRESS, 'r', newline='')
    reader = csv.reader(csvfile)
    for row in reader:
        cleanedEmailText = cleanText(row)
        cleanedEmails.append(cleanedEmailText)
    return cleanedEmails    
emails = preprocessEmails()
```

## Part B

```python
def splitData(emails, spams):
    xTrain, xTest, yTrain, yTest = train_test_split(emails, spams, test_size=0.2, random_state=42)
    return xTrain, xTest, yTrain, yTest
```

## Part C

```python
emails = preprocessEmails()

emailTexts = [email['text'] for email in emails]
emailLabels = [email['spam'] for email in emails]

xTrain, xTest, yTrain, yTest = splitData(emailTexts, emailLabels)
vectorizer = CountVectorizer()
xTrainBoW = vectorizer.fit_transform(xTrain)
xTestBoW = vectorizer.transform(xTest)
classifier = MultinomialNB()
classifier.fit(xTrainBoW, yTrain)
yPred = classifier.predict(xTestBoW)

accuracy = accuracy_score(yTest, yPred)
confMatrix = confusion_matrix(yTest, yPred)

print("Accuracy = %.4f" %accuracy)

```
